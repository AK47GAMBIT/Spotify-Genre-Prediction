```python
import pickle
with open("best_model.pkl", "rb") as f:
    best_model = pickle.load(f)

print("Model loaded successfully")
```

```python
with open("label_encoder.pkl", "rb") as f:
    le_final = pickle.load(f)

print("LabelEncoder loaded successfully")
```

```python
import pandas as pd
import numpy as np
test_df = pd.read_csv("test.csv")
test_df["energy_loudness"] = test_df["energy"] * test_df["loudness"]
test_df["tempo_energy"] = test_df["tempo"] * test_df["energy"]
test_df["dance_energy"] = test_df["danceability"] * test_df["energy"]
test_df["valence_energy"] = test_df["valence"] * test_df["energy"]
test_df["energy_to_acoustic"] = test_df["energy"] / (test_df["acousticness"] + 1e-5)
test_df["speech_to_liveness"] = test_df["speechiness"] / (test_df["liveness"] + 1e-5)
test_df["mood_score"] = test_df["valence"] * test_df["energy"]
test_df["intensity"] = test_df["energy"] * test_df["loudness"]
test_df["duration_ms"] = np.log1p(test_df["duration_ms"])
test_df["tempo"] = np.log1p(test_df["tempo"])
with open("selected_features.pkl", "rb") as f:
    selected_features = pickle.load(f)
X_test_final = test_df[selected_features]
y_test_pred = best_model.predict(X_test_final)
y_test_pred_labels = le_final.inverse_transform(y_test_pred)
test_df["predicted_genre"] = y_test_pred_labels
test_df.to_csv("test_predictions.csv", index=False)
print("Predictions saved to test_predictions.csv")
```
